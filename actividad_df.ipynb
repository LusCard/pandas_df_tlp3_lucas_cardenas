```python
import pandas as pd
import sqlite3

#Aqui agrega la conexión a la base de datos
conn = sqlite3.connect("salaries.sqlite")
#Crea el dataframe con los datos del datasets
df = pd.read_sql_query("SELECT * FROM salaries", conn)
conn.close()
df.head()
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
#todo
df.nlargest(10, 'TotalPayBenefits')[['EmployeeName', 'TotalPayBenefits']]
```

Deduje que "TotalPayBenefits" es el equivalente de "salario total(incluyendo beneficios)" así que usando el método "nlargest" hice que me de los 10 numeros más altos de TotalPayBenefits incluyendo "EmployeeName" y "TotalPayBenefits" en ese orden 

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
#todo
df['OvertimePay'] = pd.to_numeric(df['OvertimePay'])
df[df['OvertimePay'] > 50000][['EmployeeName', 'OvertimePay']]
```

Uso la función pd.to_nueric para convertir los valores texto de la columna OverTimePay a tipo numérico. Se filtran los que se les paga el tiempo extra más de 50k, selecciono mostrar sólo el nombre y el pago 

3. **Contar cuántos empleados únicos hay por año**

```python
#todo
df.groupby("Year")["EmployeeName"].nunique()
```

    Agrupo con metodo 'groupby' en la columna "Year" y aplico la función 'nunique' para contar los empleados unicos.

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
#todo
count_unique= df['JobTitle'].nunique()
most_common = df['JobTitle'].value_counts().head(5)
print(f"Total de cargos unicos: {count_unique}")
print(f"Los 5 cargos más comunes son:{most_common}")
```

Consigo los valores unicos de la columna "JobTitle" con '.nunique' y el conteo de los primeros 5 de mayor cantidad de cargos con '.value_counts().head(5)'

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
#todo
avg_salary = df.groupby('Year')['TotalPayBenefits'].mean()
print(avg_salary)
```
